```python
import os
from datetime import datetime

# 1. Create the result folder if it doesn't exist
if not os.path.exists('../result'):
    os.makedirs('../result')

# 2. Simulate a recommendation result
user_mood = "Funny 😂"
recommended_movie = "The Super Mario Bros. Movie"
timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

# 3. Save this to a log file in the result folder
with open('../result/history.txt', 'a', encoding='utf-8') as f:
    f.write(f"[{timestamp}] User felt {user_mood} -> Recommended: {recommended_movie}\n")

print("✅ Recommendation saved to result/history.txt!")
```
